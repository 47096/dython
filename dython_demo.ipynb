```python
import pandas as pd
from dython.nominal import associations, identify_nominal_columns
```

```python
# Local demo data (vendored)
df = pd.read_csv("data/pokemon.csv")
df.head()
```

```python
# Which columns are categorical?
categorical_features = identify_nominal_columns(df)
categorical_features
```

```python
# Full association matrix — this was missing before
complete_correlation = associations(df, nominal_columns=categorical_features)

df_complete_corr = complete_correlation["corr"]
df_complete_corr.dropna(axis=1, how="all").dropna(axis=0, how="all").style.background_gradient(
    cmap="coolwarm", axis=None
)
```
